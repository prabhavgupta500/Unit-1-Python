SET 3, 10 mark Question 1

```python
# A shopkeeper wants a simple billing program. Accept and item name, quantity(int) and unit price(float) from the user. Calculate the total cost, and print a neatly formatted bill using f-string alignment, showing the item name, quantity, unit price, and total, each in its own aligned column
bill={}
n= int(input("Number of Items: "))
for i in range(n):
  item= input("Enter Item Name: ")
  quantity= int(input("Enter the Quantity of the Item: "))
  price = int(input("Enter the price of the Item: "))
  bill[item]={"quantity":quantity,"price": price}
print("="*70)
print(f"{'Item':<15}{'Quantity':<15}{'Unit Price':<15}{'Total':<15}")
print("-"*70)
total_bill= 0
for item ,details in bill.items():
  quantity= int(details["quantity"])
  price= int(details["price"])
  total= quantity * price
  total_bill+=total
  print(f"{item:<15}{quantity:<15}{price:<15.2f}{total:<15.2f}")
print("="*70)
print(f"{"Total Bill":<48}{total_bill}")
```

```python
#Write a python program that accepts a list of at least 8 numbers nd finds the Second Largest number in the list without using the sorted() function.
numbers= list(map(int,input("Enter atleast 8 numbers seperated by a space: ").split()))
if len(numbers)<8:
  print("Enter Atleast 8 numbers")
else:
  largest= numbers[0]
  second_largest= numbers[0]
  for num in numbers:
    if num > largest:
      second_largest= largest
      largest= num
    elif num > second_largest and num!=largest:
      second_largest= num
print("The Second Largest Number is: ",second_largest)
```
